```python
Step 5 — Build Core Customer Features
Now create the first customer-level table.

5A — Define reference date
The final delivered purchase occurred on:
```

```python
reference_date = (
    order_analytics["order_purchase_timestamp"]
    .max()
    .normalize()
    + pd.Timedelta(days=1)
)

print("Reference date:", reference_date)
```

```python
# Aggregate core features
```

```python
customer_features = (
    order_analytics
    .groupby("customer_unique_id")
    .agg(
        first_purchase_date=(
            "order_purchase_timestamp", "min"
        ),

        last_purchase_date=(
            "order_purchase_timestamp", "max"
        ),

        frequency=(
            "order_id", "nunique"
        ),

        monetary_value=(
            "merchandise_value", "sum"
        ),

        total_order_value=(
            "order_value", "sum"
        ),

        avg_order_value=(
            "order_value", "mean"
        ),

        total_items=(
            "item_count", "sum"
        ),

        avg_items_per_order=(
            "item_count", "mean"
        )
    )
    .reset_index()
)
```

```python
# Recency and tenure

customer_features["recency_days"] = (
    reference_date
    - customer_features["last_purchase_date"]
).dt.days

customer_features["tenure_days"] = (
    reference_date
    - customer_features["first_purchase_date"]
).dt.days
```

```python
customer_features.head()
```

Validate customer grain

```python
core_feature_validation = pd.Series({
    "rows":
        len(customer_features),

    "unique_customers":
        customer_features["customer_unique_id"].nunique(),

    "duplicate_customer_ids":
        customer_features["customer_unique_id"].duplicated().sum(),

    "missing_customer_ids":
        customer_features["customer_unique_id"].isna().sum(),

    "min_frequency":
        customer_features["frequency"].min(),

    "max_frequency":
        customer_features["frequency"].max(),

    "negative_recency":
        (customer_features["recency_days"] < 0).sum(),

    "negative_tenure":
        (customer_features["tenure_days"] < 0).sum(),

    "negative_monetary":
        (customer_features["monetary_value"] < 0).sum()
})

core_feature_validation
```

```python
customer_features["purchase_span_days"] = (
    customer_features["last_purchase_date"]
    - customer_features["first_purchase_date"]
).dt.days

customer_features["is_repeat_customer"] = (
    customer_features["frequency"] > 1
).astype(int)
```

```python
# Now inspect repeat-customer prevalence:

repeat_summary = pd.Series({
    "total_customers":
        len(customer_features),

    "one_time_customers":
        (customer_features["frequency"] == 1).sum(),

    "repeat_customers":
        (customer_features["frequency"] > 1).sum(),

    "repeat_customer_pct":
        (customer_features["frequency"] > 1).mean() * 100
})

repeat_summary
```

```python
customer_freight = (
    order_analytics
    .groupby("customer_unique_id")
    .agg(
        total_freight=(
            "freight_value", "sum"
        ),

        avg_freight_per_order=(
            "freight_value", "mean"
        ),

        avg_merchandise_per_order=(
            "merchandise_value", "mean"
        )
    )
    .reset_index()
)

customer_freight.head()
```

```python
# Freight-to-merchandise ratio
# Do not average order-level freight percentages. Calculate the ratio from customer-level totals.

customer_freight["freight_to_merchandise_ratio"] = (
    customer_freight["total_freight"]
    /
    customer_features.set_index("customer_unique_id")
    .loc[
        customer_freight["customer_unique_id"],
        "monetary_value"
    ]
    .to_numpy()
)
```

```python
freight_validation = pd.Series({
    "rows":
        len(customer_freight),

    "unique_customers":
        customer_freight["customer_unique_id"].nunique(),

    "missing_total_freight":
        customer_freight["total_freight"].isna().sum(),

    "negative_total_freight":
        (customer_freight["total_freight"] < 0).sum(),

    "missing_ratio":
        customer_freight["freight_to_merchandise_ratio"].isna().sum(),

    "infinite_ratio":
        np.isinf(
            customer_freight["freight_to_merchandise_ratio"]
        ).sum(),

    "max_ratio":
        customer_freight["freight_to_merchandise_ratio"].max()
})

freight_validation
```

```python
# Join freight features


customer_features = (
    customer_features
    .merge(
        customer_freight[
            [
                "customer_unique_id",
                "total_freight",
                "avg_freight_per_order",
                "avg_merchandise_per_order",
                "freight_to_merchandise_ratio"
            ]
        ],
        on="customer_unique_id",
        how="left",
        validate="one_to_one"
    )
)
```
